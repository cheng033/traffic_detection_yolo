```python
import zipfile
import os

zip_path = "/content/data070802.zip"  # 請替換成你的 .zip 檔案路徑
extract_path = "/content/roboflow_dataset" # 請替換成你想要解壓縮到的目標路徑

# 檢查目標路徑是否存在，如果不存在則建立
if not os.path.exists(extract_path):
    os.makedirs(extract_path)
    print(f"已建立目標資料夾：{extract_path}")

# 檢查 .zip 檔案是否存在
if not os.path.exists(zip_path):
    print(f"錯誤：找不到指定的 .zip 檔案：{zip_path}")
else:
    try:
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_path)
        print(f"檔案已成功解壓縮到：{extract_path}")
    except zipfile.BadZipFile:
        print(f"錯誤：無效的 Zip 檔案：{zip_path}")
    except Exception as e:
        print(f"解壓縮時發生錯誤：{e}")
```

```python
from ultralytics import YOLO

# 使用內建模型 (你也可以換 yolov8s.pt 或 yolov8m.pt)
model = YOLO("yolov8x.pt")

# 開始訓練
model.train(data="/content/roboflow_dataset/data.yaml", epochs=50, imgsz=640, batch=16)
model.save("modelv5x.pt")
```
