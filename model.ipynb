```python
import zipfile
import os
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt
import glob
```

```python
# 1. 解壓縮
zip_path = "data0708.zip"
extract_path = "roboflow_dataset"
if not os.path.exists(extract_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
```

```python
# 2. 訓練
model = YOLO("yolov8s.pt")
model.train(data="roboflow_dataset/data.yaml", epochs=100, imgsz=640)
model.save("modelv3.pt")
```

```python
# 3. 顯示訓練結果
result_img = "runs/detect/train/results.png"
if os.path.exists(result_img):
    img = Image.open(result_img)
    plt.imshow(img)
    plt.axis('off')
    plt.show()
```

```python
# 4. 預測
model = YOLO("modelv3.pt")
results = model.predict(source="image.png", conf=0.3, save=True)

predict_dirs = sorted(glob.glob("runs/detect/predict*"), key=os.path.getmtime)
latest_dir = predict_dirs[-1]
img_path = glob.glob(f"{latest_dir}/*.jpg") + glob.glob(f"{latest_dir}/*.png")
if img_path:
    img = Image.open(img_path[0])
    plt.imshow(img)
    plt.axis('off')
    plt.show()
else:
    print("找不到預測結果圖片")
```
